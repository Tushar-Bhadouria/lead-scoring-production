```python
hyper_parameter =[dpa,kpa,rfpa,gbpa,xgpa]
```

```python
best_params_list = []
best_estimators = {}

# Combine models and their parameter grids
models = [DT, knn, rf, gb, xgb]
param_grids = [dpa, kpa, rfpa, gbpa, xgpa] # Ensure order matches 'models'
model_names = ['Decision Tree', 'KNearestNeighbors', 'RandomForest', 'GradientBoost', 'XGBoost']

for i, (model, param_grid) in enumerate(zip(models, param_grids)):
  print(f"\n--- Running GridSearchCV for {model_names[i]} ---")
  try:
    # Initialize GridSearchCV
    current_param_grid = param_grid.copy()

    gs = GridSearchCV(estimator=model, param_grid=current_param_grid,scoring='f1', n_jobs=-1, cv=5, verbose=1)
    gs.fit(X_train, y_train)

    # Store the best parameters
    best_params_list.append({model_names[i]: gs.best_params_})
    best_estimators[model_names[i]] = gs.best_estimator_
    print(f"Best parameters for {model_names[i]}: {gs.best_params_}")
  except Exception as e:
    print(f"Error during GridSearchCV for {model_names[i]}: {e}")
    best_params_list.append({model_names[i]: f"Error: {e}"})

print("\n--- All Best Parameters ---")
for params in best_params_list:
    print(params)
```

### **5.2 Stacking Classifier**

```python
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression

# Define base estimators using the best models identified from GridSearchCV
# Based on user's summary, Gradient Boosting, Random Forest, and XGBoost performed well.
estimators = [
    ('dt', best_estimators['Decision Tree']),
    ('rf', best_estimators['RandomForest']),
    ('xgb', best_estimators['XGBoost'])
]

# Define the meta-estimator (final estimator)
# A simple Logistic Regression is often a good choice for the meta-model
final_estimator = best_estimators['GradientBoost']

# Create the StackingClassifier
stk_classifier = StackingClassifier(
    estimators=estimators,
    final_estimator=final_estimator,
    cv=5, # Number of cross-validation folds for the base models
    n_jobs=-1 # Use all available cores
)

print("Stacking Classifier initialized with Decision Tree, Random Forest, and XGBoost as base estimators.")
```

### **5.4 ROC AUC, PR AUC and Optimal Threshold Analysis**

```python
from sklearn.metrics import roc_curve, auc, precision_recall_curve, average_precision_score
import matplotlib.pyplot as plt
import numpy as np

# Get prediction probabilities for the positive class (class 1) from the Stacking Classifier
y_pred_proba_stk = stk_classifier.predict_proba(X_test)[:, 1]

# --- ROC Curve and AUC ---
fpr, tpr, thresholds_roc = roc_curve(y_test, y_pred_proba_stk)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(10, 6))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve - Stacking Classifier')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

print(f"ROC AUC for Stacking Classifier: {roc_auc:.2f}")
```

```python
# --- Precision-Recall Curve and AUC ---
precision, recall, thresholds_pr = precision_recall_curve(y_test, y_pred_proba_stk)
avg_precision = average_precision_score(y_test, y_pred_proba_stk)

plt.figure(figsize=(10, 6))
plt.plot(recall, precision, color='green', lw=2, label=f'PR curve (area = {avg_precision:.2f})')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve - Stacking Classifier')
plt.legend(loc='lower left')
plt.grid(True)
plt.show()

print(f"Average Precision Score (PR AUC) for Stacking Classifier: {avg_precision:.2f}")
```

#### Optimal Classification Threshold

We can determine an optimal classification threshold. A common method for ROC curves is Youden's J statistic, which maximizes the difference between TPR and FPR. For Precision-Recall curves, maximizing the F1-score is often used.

Let's find the optimal threshold using Youden's J statistic for the ROC curve, as it balances sensitivity and specificity.

```python
# Calculate Youden's J statistic for each threshold
youdens_j = tpr - fpr
optimal_idx_roc = np.argmax(youdens_j)
optimal_threshold_roc = thresholds_roc[optimal_idx_roc]

print(f"Optimal threshold (Youden's J) from ROC curve: {optimal_threshold_roc:.4f}")

# Alternatively, find the threshold that maximizes F1-score on the PR curve
# Note: thresholds_pr has one less element than precision and recall, so align indices
f_scores = (2 * precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1])
# Handle potential division by zero if precision + recall is zero
f_scores[np.isnan(f_scores)] = 0
optimal_idx_pr = np.argmax(f_scores)
optimal_threshold_pr = thresholds_pr[optimal_idx_pr]

print(f"Optimal threshold (max F1-score) from PR curve: {optimal_threshold_pr:.4f}")

# Let's use the F1-score optimized threshold for re-evaluation
new_threshold = optimal_threshold_pr

# Apply the new threshold to make predictions
y_pred_stk_new_threshold = (y_pred_proba_stk >= new_threshold).astype(int)

# Re-evaluate metrics with the new threshold
accuracy_stk_new = accuracy_score(y_test, y_pred_stk_new_threshold) * 100
precision_stk_new = precision_score(y_test, y_pred_stk_new_threshold) * 100
recall_stk_new = recall_score(y_test, y_pred_stk_new_threshold) * 100
f1_stk_new = f1_score(y_test, y_pred_stk_new_threshold) * 100

print(f"\nStacking Classifier Performance with new threshold ({new_threshold:.4f}):")
print(f"Accuracy: {accuracy_stk_new:.2f}%")
print(f"Precision: {precision_stk_new:.2f}%")
print(f"Recall: {recall_stk_new:.2f}%")
print(f"F1-score: {f1_stk_new:.2f}%")

# You might want to update the results DataFrame with this new performance if desired.
```
